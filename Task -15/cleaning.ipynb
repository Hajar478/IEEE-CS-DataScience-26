### Spoting specific values

converted any "UNKNOWN" or "ERROR" values into np.nan 
by converting df into array and perform boolean mask 

```python
data_arr = mydata.to_numpy()
```

```python
print(data_arr)
```

```python
mask = (data_arr==['UNKNOWN' and 'ERROR'])


data_arr[mask]=np.nan
```

```python
print(data_arr)
```

```python
mydata_new = pd.DataFrame(data_arr,columns=mydata.columns ,index=mydata.index)
```
