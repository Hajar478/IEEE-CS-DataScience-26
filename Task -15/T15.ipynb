### Step 7: Handle Missing Values (Part 2: Fill NaNs)

##### fill missing `Price Per Unit`

```python
df['Price Per Unit']=df['Price Per Unit'].fillna(0)
```

```python
print(df['Price Per Unit'].isnull().sum())
```

##### handle other missing values appropriately (like Payment Method or Location)

```python
mask2 =(df['Total Spent']<=0) & (df['Payment Method'].isna())

df['Payment Method']= np.where(mask2,df['Payment Method'].fillna('not payed'),df['Payment Method'].fillna('other'))

```

```python
df['Location']=df['Location'].fillna('other')
```

```python
print(df[['Payment Method','Location']])
```

### Step 8: Logical Validation

##### create a new column `Expected_Total` by multiplying `Quantity * Price Per Unit`

```python
df['Expected_Total']=(df['Quantity']*df['Price Per Unit'])
```

```python
df
```

##### filter rows where Total Spent is NOT equal to Expected_Total

```python
df['Total Spent'] = np.where(df['Total Spent'] != df['Expected_Total'], np.nan, df['Total Spent'])
```

```python
print(df['Total Spent'])
```

##### fill missing or incorrect Total Spent values using the Expected_Total calculation.

```python
df['Total Spent'] = df['Total Spent'].fillna(df['Expected_Total'])
```

```python
print(df[['Total Spent','Expected_Total']])
```
