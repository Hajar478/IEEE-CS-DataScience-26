### Step 1. Import the necessary libraries:

```python
import pandas as pd 
import matplotlib as plt
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np 
import datetime as dt
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv). 

### Step 3. Assign it to a variable called tips

```python
tips= pd.read_csv("https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/07_Visualization/Tips/tips.csv")
```

```python
tips.info()
```

### Step 4. Delete the Unnamed 0 column

```python
tips = tips.loc[:, ~tips.columns.str.contains('^Unnamed')]
```

```python
tips.head(10)
```

```python
tips['total_bill'].max()
```

### Step 5. Plot the total_bill column histogram

```python
plt.style.use('fivethirtyeight')

tips['total_bill'].hist()
plt.xlim(0,60)
plt.ylim(0,80)
plt.title('Total Bill ')
plt.grid(False)
plt.show()
```

### Step 6. Create a scatter plot presenting the relationship between total_bill and tip

```python
plt.figure(figsize=(10,7))


sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip',
    alpha=0.6,s=100
)
plt.xlabel=('Total Bill')
plt.ylabel=('Tip')
plt.grid(False)
plt.title('Total bill vs Tips')
plt.show()
```

### Step 7.  Create one image with the relationship of total_bill, tip and size.
#### Hint: It is just one function.

```python
sns.pairplot(
    tips,
    vars=["total_bill", "tip", "size"],
    hue="size",
    palette="viridis",
    diag_kind="kde" 
)


plt.suptitle("Relationship between Total Bill, Tip, and Size", y=1.02)
plt.show()
```

### Step 8. Present the relationship between days and total_bill value

```python
sns.barplot(
    data=tips,
    x='day',
    y='total_bill',
    hue='day',
    palette="Blues"
)
plt.ylabel=('Total Bill')
plt.xlabel=('Day')
plt.grid(False)
plt.title('Total bill per day')
plt.show()
```

### Step 9. Create a scatter plot with the day as the y-axis and tip as the x-axis, differ the dots by sex

```python
plt.figure(figsize=(10,6))

sns.scatterplot(
    data=tips,
    x='tip',
    y='day',
    hue='sex',
    alpha=0.6,s=100
)
plt.xlabel=('Tip')
plt.ylabel=('day')
plt.grid(False)
plt.title('Tips vs days')
plt.show()
```

### Step 10.  Create a box plot presenting the total_bill per day differetiation the time (Dinner or Lunch)

```python
tips
```

```python
sns.boxplot(
    data=tips,
    x='day',
    y='total_bill',
    hue='time',
    hue_order=['Dinner','Lunch']
)
plt.legend(loc='upper right')
```

### Step 11. Create two histograms of the tip value based for Dinner and Lunch. They must be side by side.

```python
sns.displot(
    data=tips,
    x='tip',
    col='time',
    kde=True
)
```

### Step 12. Create two scatterplots graphs, one for Male and another for Female, presenting the total_bill value and tip relationship, differing by smoker or no smoker
### They must be side by side.

```python
sns.relplot(
    data=tips,
    x="total_bill",
    y='tip',
    col="sex",
    hue="smoker",
    kind="scatter"
)
```
