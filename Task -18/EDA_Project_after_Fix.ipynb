### 7-(sales vs rating together)

the correlation number (0.30) only tells us rating and sales are weakly linked butttt it doesnt answer the actual question the firm is asking: should they invest based on sales,rating,or both?

so i splited the games into 4 groups using the median(median isn't really affected by outliers) sales and median rating as cut points, then i compare the average sales of each group. this directly answers the business question.

   

```python
med_sales = merged['metrics_sales'].median()
med_rating = merged['meta_score'].median()

print('median sales:', med_sales)
print('median rating:', med_rating)
```

```python

# i made this function to filter each game into one of 4 groups 

def get_quadrant(row):
    if pd.isna(row['meta_score']):
        return np.nan
    if row['metrics_sales']>=med_sales and row['meta_score']>=med_rating:
        return 'High Sales & High Rating'
    elif row['metrics_sales']>=med_sales and row['meta_score']<med_rating:
        return 'High Sales & Low Rating'
    elif row['metrics_sales']<med_sales and row['meta_score']>=med_rating:
        return 'Low Sales & High Rating'
    else:
        return 'Low Sales & Low Rating'

merged['quadrant'] = merged.apply(get_quadrant, axis=1)
```

```python
quadrant_summary = merged.groupby('quadrant').agg(
    game_count=('title','count'),
    avg_sales=('metrics_sales','mean'),
    avg_rating=('meta_score','mean')
).round(2)

quadrant_summary
```

```python
# scatter plot to see the 4 groups visually
plt.figure(figsize=(10,6))

sns.scatterplot(data=merged, x='meta_score', y='metrics_sales', hue='quadrant', alpha=0.6)
plt.axvline(med_rating, color='gray', linestyle='--')
plt.axhline(med_sales, color='gray', linestyle='--')
plt.yscale('log')
plt.title('Sales vs Rating - 4 Investment Groups')
plt.xlabel('Meta Score (critic rating)')
plt.ylabel('Sales in Millions (log scale)')

plt.show()
```

```python
# High Sales & High Rating games sell almost double the High Sales & Low Rating games on average
# this group is the smallest risk for the firm, good reviews & proven high sales at the same time
# Low Sales & High Rating games are the riskiest to judge by rating alone ,its liked but no much sales
# so rating alone is NOT a safe investment signal, but rating boosts sales when it's already a high-selling game
```

### 8- Final Report

**Answering the business question: should the firm invest based on sales, ratings, or both?**

1- Rating alone is not enough. The correlation between meta_score and sales is weak (~0.3), and the 'Low Sales & High Rating' group proves that plenty of critically praised games still sell poorly. A firm that only invests in high-rated games can still lose money.

2- Sales alone is not enough either. The 'High Sales & Low Rating' group still exists (games that sold well without great reviews زي جيمز من فيلم مشهور)but they sell on average almost half of what the 'High Sales & High Rating' group sells, and they carry more reputational/franchise-decline risk long term

3- The safest and most profitable group is **High Sales & High Rating**
 it is the largest group in the data and has the highest average sales. Good reviews on top of an already strong seller is the closest thing to a safe bet in this data

4- Recommendation: the firm should prioritize games that show early signs of both good reviews & strong sales potential (like proven high-rated franchises or popular genres from the genre analysis on the platforms with the highest total sales).

Rating should be used as a supporting signal to protect long term brand value (like a bonus), not as the main investment driver, and sales history/franchise strength should be used to avoid pure 'critical darling, commercial flop' risk.
