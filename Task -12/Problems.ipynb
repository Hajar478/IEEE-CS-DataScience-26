2. Create a 7x7 matrix of all ones Inside it place a 5x5 "hole" of zeros (from the 2nd to the 6th row/column)

Finally place the number 5 exactly in the center of the matrix



```python
arr1 = np.ones((7,7))

hole = np.zeros((5,5))
hole[2,2] = 5

arr1[1:-1 , 1:-1] = hole

print(arr1)
```

3. Ask the user for a number n. Then, use broadcasting to create a n x n np array that has its first column full of ones, its second column full of twos, its third column full of threes, etc…

```python
n=int(input())

nxn = np.zeros((n,n))
cols=0
for x in range(n):
    nxn[:,cols] = x+1
    cols +=1


print(nxn)
```

6. You have a 1D array representing student scores: scores = np.array([45, 88, 92, 30, 50, 76, 99, 12, 65])

1- Use Boolean Indexing to filter and print only the passing grades (>= 50)

2- check if any student got a perfect score of 100

3- check if all students passed



```python
scores = np.array([45, 88, 92, 30, 50, 76, 99, 12, 65])

'''mask = scores>=50        #boolean masking
filtered = scores[mask]
print(f'Passing : {filtered}')'''

filtered=[]               #boolean index filtering
for score in scores:
    if score>=50:
        filtered.append(True)
    else:
        filtered.append(False)

print(f" passing : {scores[filtered]}")



'''if 100  in scores:
   print("Perfect score achieved")
else:
    print('Perfect score not achieved')'''

if scores.any() == 100:
    print("Perfect score achieved")
else :
    print("Perfect score not achieved")



if scores.all() >= 50:
    print("all pased")
else :
    print("There is faliers")


```
