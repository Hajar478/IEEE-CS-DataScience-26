### Resampleing

with using Freq

D	تقويم يومي (Calendar day)
W	أسبوعي (Weekly)
M	نهاية الشهر (Month end)
Q	ربع سنوي (Quarter end)

```python
df.resample("M")
```

### Aggregate

helps in multiple calculations at once
cane be used with resample for weekly ,monthly or even yearly report analysis

```python
df.resample("W").agg({'Close': "mean",'High':'max'})
```

```python
# fist droping them
df.dropna(how="any").shape


# dropping any row with NaN value and checking the shape to see how much rows dropped
```

```python
df.dropna(subset=["Close","Low"],how="any").shape
#if any row in these cols is nan drop it
```

```python
df.dropna(subset=["Close","Low"],how="any").shape
#if any row in these cols together is nan drop it
```
